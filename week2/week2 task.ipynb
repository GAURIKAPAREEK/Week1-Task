```python
# step 1 .Load dataset
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv('heels.csv')
```

```python
df.head()
```

```python
conn = sqlite3.connect('heels_sales.db')
```

```python
df.to_sql('heels', conn, if_exists='replace', index=False)
```

```python
# step 2  schema
query = "PRAGMA table_info(heels);"
pd.read_sql(query, conn)
```

```python
 # step 2 sample data 
query = "SELECT * FROM heels LIMIT 5"
pd.read_sql(query, conn)
```

```python

query="SELECT COUNT(*) AS total_rows FROM heels";
pd.read_sql(query, conn)
```

```python
query="SELECT category, COUNT(*) AS product_count FROM heels GROUP BY category";
pd.read_sql(query, conn)
```

```python
# # step 3 .Apply WHERE filters (rating)
query="SELECT product_id, title, rating  FROM heels  WHERE rating > 4";
pd.read_sql(query, conn)
```

```python
# discount
query="SELECT title, discount FROM heels WHERE discount > 50";
pd.read_sql(query, conn)
```

```python
# category
query="SELECT title, category FROM heels WHERE category = 'heels'";
pd.read_sql(query, conn)
```

```python
#initial price
query="SELECT title, initial_price  FROM heels WHERE initial_price > 3000";
pd.read_sql(query, conn)
```

```python
# step 4 aggregations 
query="SELECT AVG(rating) AS average_rating FROM heels";
pd.read_sql(query, conn)
```

```python
# count
query="SELECT SUM(ratings_count) AS total_ratings FROM heels";
pd.read_sql(query, conn)
```

```python
# group by category and averge
query="SELECT category, AVG(initial_price) AS avg_price FROM heels GROUP BY category";
pd.read_sql(query, conn)
```

```python
# step 5 top category product having highest rating
query="SELECT title, rating FROM heels ORDER BY rating DESC LIMIT 1";
pd.read_sql(query, conn)
```

```python
# Top 3 Most Expensive Products
query="SELECT title, initial_price FROM heels ORDER BY initial_price DESC LIMIT 3;"
pd.read_sql(query, conn)
```

```python
# top 3 rated products
query="SELECT title, ratings_count FROM heels ORDER BY ratings_count DESC LIMIT 3";
pd.read_sql(query, conn)
```

```python
# Step 6 Use Cases

# Products with Highest Discount
query="SELECT title, discount FROM heels ORDER BY discount DESC LIMIT 5";
pd.read_sql(query, conn)
```

```python
# Products with No Ratings
query="SELECT title, rating, ratings_count FROM heels WHERE ratings_count = 0";
pd.read_sql(query, conn)
```

```python
# duplicates count
query="SELECT title, COUNT(*) AS duplicate_count FROM heels GROUP BY title HAVING COUNT(*) > 1";
pd.read_sql(query, conn)
```

```python
# Step 7: Data Validation and Quality Checks
# check for null values
query="SELECT * FROM heels WHERE rating IS NULL OR initial_price IS NULL OR category IS NULL";
pd.read_sql(query, conn)
```

```python
# Count Missing Discounts
query="SELECT COUNT(*) AS missing_discount FROM heels WHERE discount IS NULL";
pd.read_sql(query, conn)
```
