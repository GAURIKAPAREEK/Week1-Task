```python
#This file only contains Practical Implementations
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("dataFrame").getOrCreate()
```

```python
# Step 1 Reading CSV file
df = spark.read.csv("data/source.txt", header=True, inferSchema=True)
```

```python
df.show()
```

```python
# Count total rows
df.count()
```

```python
#Print Schema
df.printSchema()
```

```python
#Creating and writing data in Parquet File Format
data = [
    (1, "Laptop", 55000),
    (2, "Mouse", 800),
    (3, "Keyboard", 1500)
]
dff = spark.createDataFrame(data, ["id", "product", "price"])
dff.write.mode("overwrite").parquet("product.parquet")
```

```python
#Reading Parquet file
parquet_df = spark.read.parquet("product.parquet")
parquet_df.show()
```

```python
#Print Schema
parquet_df.printSchema()
```

```python
#Step 2 Filtering and Selecting Data
dff=df.select("user_id", "base_price","category").filter(df.category == "Electronics")
dff.show()
```

```python
#Filtering and Selecting Data
newdf=df.filter((df.status == "Completed") & (df.amount > 1000))
newdf.show()
```

```python
#Step 3 Transfromations 

#Renaming Column
df1 = df.withColumnRenamed("product_price","price",)
```

```python
df1.show()
```

```python
df1.printSchema()
```

```python
#Changing Data Type
from pyspark.sql.functions import col
df1 = df1.withColumn("price", col("price").cast("string"))
```

```python
df1.printSchema()
```

```python
from pyspark.sql.functions import col
df1 = df1.withColumn("price", col("price").cast("double"))
```

```python
df1.printSchema()
```

```python
#Creating new column
from pyspark.sql.functions import col
df1 = df1.withColumn("final_price", col("base_price") * 1.18)
df1.show()
```

```python
#Step 4 Handling null values (Dropping it)
clean_df = df1.dropna()
clean_df.show()
```

```python
#Step 5 wide transformations
from pyspark.sql.functions import sum
result_df = df1.groupBy("category").agg(sum("price").alias("total_price"))
result_df.show()
```

```python
#Predicate Pushdown
from pyspark.sql.functions import col 
filtered_df = df1.filter(col("price") > 1000) 
filtered_df.show()
```

```python
#Step 6 Building Pipeline
from pyspark.sql.functions import col
pipeline = (spark.read.csv("data/source.txt", header=True, inferSchema=True).withColumnRenamed("product_price", "price").withColumn("price", col("price").cast("double")).withColumn("final_price", col("base_price") * 1.18).filter(col("category") == "Electronics"))
pipeline.show(5)
```

```python
#Step 7 Save processed data into CSV Format
pipeline.write.mode("overwrite").option("header", True).csv("output_pipeline")
```
